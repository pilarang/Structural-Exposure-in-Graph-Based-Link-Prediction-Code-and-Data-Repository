# Step 3. NOO Filtering authorships para datos al 2024

Paso 2 de la implementación en la BDUNAM.

En este paso se utiliza el archivo generado en el paso anterior *nodes_catalogue.csv* NO VAMOS A para filtrar y mantener solo los autores que han escrito 3 o más artículos.

Se filtran las dos bases obtenidas en el paso anterior.
El conjunto entrenamiento 2000 a 2020 cuenta con 2788 (ANTES: 1981-2017=1943 registros, tras el filtrado se queda con 1671).
El conjunto de validación 2021 a 2022 cuenta con 1572 (ANTES 2018-2021= 650 registros, tras el filtrado se queda con 545)
AHORA se incorpora
El conjunto de test de 2023-2024 cuenta con 895 registros

ESTOS CONJUNTOS DE DATOS SON SOLO PARA NODOS CONECTADOS, SE TIENE QUE BALANCEAR INCORPORANDO DATOS NO CONECTADOS

Los archivos filtrados se almacenan en otros CSV llamados *UNAM_Completo_filtered_1981_2017.csv* y *UNAM_Completo_filtered_2018_2021.csv*

```python
import pandas as pd
```

```python
#Leer el archivo generado anteriormente
df = pd.read_csv('nodes_catalogue.csv')
```

```python
df
```

```python
#lista de autores
autores = df['Autor_norm'].to_list()
```

## Filtrar autores

```python
df_original = pd.read_csv('UNAM_Completo_Corregido.csv')
```

```python
#Mostrar dimension de Data Frame inicial
df_original.shape
```

```python
df_2000_2020 = pd.read_csv('UNAM_training_2000_2020.csv')
df_2021_2022 = pd.read_csv('UNAM_val_2021_2022.csv')
df_2023_2024 = pd.read_csv('UNAM_test_2023_2024.csv')
```

```python
df_2000_2020.shape, df_2021_2022.shape, df_2023_2024.shape
```

### Datos de 2000-2020

```python
#Seleccionar autores que se encuentren en la lista de autores
df_2000_2020 = df_2000_2020[df_2000_2020['Autor_norm'].isin(autores)]
```

```python
#Mostrar dimension de Data Frame despues del filtrado-- SON LOS MISMOS!
df_2000_2020.shape
```

```python
df_2000_2020 = df_2000_2020.reset_index(drop = True)
```

```python
df_2000_2020.to_csv('UNAM_training_2000_2020.csv',index=False)
```

### Datos de 2021-2022

```python
#Seleccionar autores que se encuentren en la lista de autores
df_2021_2022 = df_2021_2022[df_2021_2022['Autor_norm'].isin(autores)]
```

```python
#Mostrar dimension de Data Frame despues del filtrado
df_2021_2022.shape
```

```python
df_2021_2022 = df_2021_2022.reset_index(drop = True)
```

```python
df_2021_2022.to_csv('UNAM_val_2021_2022.csv',index=False)
```

### Datos de 2023-2024

```python
#Seleccionar autores que se encuentren en la lista de autores
df_2023_2024 = df_2023_2024[df_2023_2024['Autor_norm'].isin(autores)]
```

```python
#Mostrar dimension de Data Frame despues del filtrado
df_2023_2024.shape
```

```python
df_2023_2024 = df_2023_2024.reset_index(drop = True)
```

```python
df_2023_2024.to_csv('UNAM_test_2023_2024.csv',index=False)
```
