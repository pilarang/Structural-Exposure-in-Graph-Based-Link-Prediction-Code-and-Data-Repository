Segundo, se selecciona el titulo y el autor, y se cuenta el número de veces que aparece cada autor (el numero de artículos que ha escrito). El resultado se almacena en *nodes_catalogue.csv*, y contiene el nombre de cada autor y el número de artículos que ha escrito.

## Separar Base de Datos por fechas

70

Datos para entrenamiento: 2000 - 2022 70%  

Datos para validación: 2023   15 %

Datos para test: 2024  15%

```python
#número de artículos por año
#df_original['Año'].value_counts()
```

```python
#CSV con los datos de entrenamiento desde 2000 a 2020 
df_2000_2020 = df_original[df_original['Año'] < 2021]
df_2000_2020.reset_index(drop = True)
df_2000_2020.to_csv('UNAM_training_2000_2020.csv',index=False)
df_2000_2020.shape
```

```python
2788/5255 
#53% para entrenamiento
```

```python
#CSV con los datos de validacion desde 2019 a 2021 si fuera 15% deberia ser 
df_2021_2022 = df_original[(df_original['Año'] >= 2021) & (df_original['Año'] < 2023)] 
df_2021_2022.reset_index(drop=True, inplace=True)
df_2021_2022.to_csv('UNAM_val_2021_2022.csv', index=False)
print(df_2021_2022.shape)
```

```python
1572/5255
# 30% de validación
```

```python
#CSV con los datos de test desde 2021 en adelante (2024)
df_2023_2024 = df_original[df_original['Año'] >= 2023]
df_2023_2024.reset_index(drop = True)
df_2023_2024.to_csv('UNAM_test_2023_2024.csv',index=False)
df_2023_2024.shape
```

```python
895/5255
# 17% de test
```
